# Coordinate transformation and CRS units

A coordinate transform converts coordinate values. It does not establish that the destination CRS is appropriate for every analysis. This example is geographically limited to central Florida.


## Learning objective
Use a small, deterministic example to verify spatial behavior and make assumptions explicit.


```python
from qgis.core import QgsCoordinateReferenceSystem, QgsCoordinateTransform, QgsGeometry, QgsPointXY, QgsProject

source = QgsCoordinateReferenceSystem('EPSG:4326')
target = QgsCoordinateReferenceSystem('EPSG:32617')
assert source.isValid() and target.isValid()
point = QgsGeometry.fromPointXY(QgsPointXY(-81.0, 28.5))
transform = QgsCoordinateTransform(source, target, QgsProject.instance())
point.transform(transform)
x, y = point.asPoint().x(), point.asPoint().y()
assert 0 < x < 1_000_000 and 0 < y < 10_000_000
print(f'Projected coordinates in {target.authid()}: {x:.2f}, {y:.2f}')

# Choose a local projected CRS for measurement and document its units.
```
