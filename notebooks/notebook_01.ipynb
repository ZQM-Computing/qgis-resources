# Vector layers and synthetic features

Create a memory point layer, add two synthetic features, and assert its schema and feature count. Run in a QGIS-enabled Python kernel.


## Learning objective
Use a small, deterministic example to verify spatial behavior and make assumptions explicit.


```python
from qgis.core import QgsFeature, QgsGeometry, QgsPointXY, QgsVectorLayer
from qgis.PyQt.QtCore import QVariant

sites = QgsVectorLayer('Point?crs=EPSG:3857', 'sites', 'memory')
sites.dataProvider().addAttributes([])
sites.updateFields()
features = []
for x, y in [(10, 20), (30, 40)]:
    feature = QgsFeature(sites.fields())
    feature.setGeometry(QgsGeometry.fromPointXY(QgsPointXY(x, y)))
    features.append(feature)
ok, added = sites.dataProvider().addFeatures(features)
sites.updateExtents()
assert ok and sites.isValid() and sites.featureCount() == 2
print('CRS:', sites.crs().authid(), '| feature count:', sites.featureCount())

# Memory layers are ephemeral; export to GeoPackage for durable outputs.
```
