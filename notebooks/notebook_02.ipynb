# Geometry validity and spatial predicates

Use explicit synthetic geometries to test intersection and containment. Boundary semantics matter: a point on a polygon edge intersects the polygon but is not within its interior.


## Learning objective
Use a small, deterministic example to verify spatial behavior and make assumptions explicit.


```python
from qgis.core import QgsGeometry, QgsPointXY

zone = QgsGeometry.fromWkt('POLYGON((0 0,100 0,100 100,0 100,0 0))')
inside = QgsGeometry.fromPointXY(QgsPointXY(25, 25))
boundary = QgsGeometry.fromPointXY(QgsPointXY(0, 50))
outside = QgsGeometry.fromPointXY(QgsPointXY(150, 150))
assert zone.isGeosValid()
assert inside.within(zone)
assert boundary.intersects(zone)
assert not boundary.within(zone)
assert not outside.intersects(zone)
print('Interior, boundary, and exterior cases passed')

# If business rules require exclusive zone assignment, define how boundary points are resolved.
```
